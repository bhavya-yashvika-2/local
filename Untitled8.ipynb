```python
import sqlite3
import pandas as pd
from datetime import date
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch

# ------------------ DATABASE CONNECTION ------------------

try:
    db = sqlite3.connect("milk_dairy.db")
    cursor = db.cursor()
except Exception as err:
    print("Database Connection Failed:", err)
    exit()

# ------------------ CREATE TABLES IF NOT EXISTS ------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    phone TEXT,
    rate REAL NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS milk_records (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER,
    date TEXT,
    morning REAL,
    evening REAL,
    total REAL,
    FOREIGN KEY(customer_id) REFERENCES customers(id)
)
""")

db.commit()

# ------------------ REGISTER CUSTOMER ------------------

def register_customer():
    try:
        name = input("Enter Name: ")
        phone = input("Enter Phone: ")
        rate = float(input("Enter Rate per Liter: "))

        sql = "INSERT INTO customers (name, phone, rate) VALUES (?,?,?)"
        cursor.execute(sql, (name, phone, rate))
        db.commit()

        print("✅ Customer Registered Successfully!")

    except Exception as e:
        print("Error:", e)

# ------------------ ADD DAILY MILK ENTRY ------------------

def add_milk_entry():
    try:
        cust_id = int(input("Enter Customer ID: "))

        cursor.execute("SELECT rate FROM customers WHERE id=?", (cust_id,))
        result = cursor.fetchone()

        if result is None:
            print("❌ Customer Not Found!")
            return

        morning = float(input("Morning Milk (liters): "))
        evening = float(input("Evening Milk (liters): "))

        total = morning + evening
        today = date.today().strftime("%Y-%m-%d")

        sql = """INSERT INTO milk_records 
                 (customer_id, date, morning, evening, total) 
                 VALUES (?,?,?,?,?)"""

        cursor.execute(sql, (cust_id, today, morning, evening, total))
        db.commit()

        print("✅ Milk Entry Added!")

    except Exception as e:
        print("Error:", e)

# ------------------ MONTHLY BILL ------------------

def monthly_bill():
    try:
        cust_id = int(input("Enter Customer ID: "))
        month = input("Enter Month (YYYY-MM): ")

        cursor.execute("SELECT name, rate FROM customers WHERE id=?", (cust_id,))
        customer = cursor.fetchone()

        if customer is None:
            print("❌ Customer Not Found!")
            return

        sql = """
        SELECT SUM(total) FROM milk_records
        WHERE customer_id = ? 
        AND strftime('%Y-%m', date) = ?
        """

        cursor.execute(sql, (cust_id, month))
        result = cursor.fetchone()

        total_liters = result[0] if result[0] else 0

        if total_liters == 0:
            print("❌ No Records Found For This Month!")
            return

        total_amount = total_liters * customer[1]

        print("\n===== MONTHLY BILL =====")
        print("Customer:", customer[0])
        print("Total Liters:", total_liters)
        print("Rate:", customer[1])
        print("Total Amount: ₹", total_amount)

        generate_pdf(customer[0], month, total_liters, customer[1], total_amount)
        export_excel(cust_id, month)

    except Exception as e:
        print("Error:", e)

# ------------------ EXPORT TO EXCEL ------------------

def export_excel(cust_id, month):
    query = """
    SELECT date, morning, evening, total 
    FROM milk_records
    WHERE customer_id=? 
    AND strftime('%Y-%m', date)=?
    """

    df = pd.read_sql_query(query, db, params=(cust_id, month))

    file_name = f"Customer_{cust_id}_{month}.xlsx"
    df.to_excel(file_name, index=False)

    print("✅ Excel File Generated:", file_name)

# ------------------ GENERATE PDF ------------------

def generate_pdf(name, month, liters, rate, amount):
    file_name = f"{name}_{month}_Bill.pdf"
    doc = SimpleDocTemplate(file_name)
    elements = []

    styles = getSampleStyleSheet()

    elements.append(Paragraph("Milk Dairy Monthly Bill", styles['Title']))
    elements.append(Spacer(1, 0.3 * inch))

    elements.append(Paragraph(f"Customer Name: {name}", styles['Normal']))
    elements.append(Paragraph(f"Month: {month}", styles['Normal']))
    elements.append(Paragraph(f"Total Liters: {liters}", styles['Normal']))
    elements.append(Paragraph(f"Rate per Liter: ₹ {rate}", styles['Normal']))
    elements.append(Paragraph(f"Total Amount: ₹ {amount}", styles['Normal']))

    doc.build(elements)

    print("✅ PDF Bill Generated:", file_name)

# ------------------ MENU ------------------

while True:
    print("\n===== MILK DAIRY SYSTEM =====")
    print("1. Register Customer")
    print("2. Add Milk Entry")
    print("3. Generate Monthly Bill")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        register_customer()
    elif choice == "2":
        add_milk_entry()
    elif choice == "3":
        monthly_bill()
    elif choice == "4":
        print("Closing System...")
        break
    else:
        print("Invalid choice!")

cursor.close()
db.close()
```
